```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[5,6,8,10,15]
x1=[2,4,6,8,10]
y1=[4,8,12,16,20]
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.xlim(1,10)
plt.ylim(5,15)
plt.title("Line Plot Example")
plt.plot(x,y,label="line-1",color='green',linestyle='dotted',linewidth=4,marker='D',markerfacecolor='purple',markersize=12)
plt.plot(x1,y1,label="line-2",color='brown',linestyle='dashed',linewidth=4,marker='*',markerfacecolor='orange',markersize=12,)
plt.legend()
plt.savefig("sample.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
Students=["Padma","Priya","Padmapriya","Ravi","Easwari"]
Marks=[66,86,96,99,66]
plt.bar(Students,Marks,color="green",width=0.6,edgecolor="red",label="StudentsMarks")
plt.legend()
plt.title("Students Marks In Semester")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.savefig("sample1.png")
plt.show()
```

```python
hours=[2,3,4,5,6]
marks=[45,52,86,96,99]
plt.scatter(hours,marks,color="brown",marker="*",s=160,label="Studentmarks")
plt.legend()
plt.xlabel("hours")
plt.ylabel("marks")
plt.title("Student Studies")
plt.savefig("sample2.png")
plt.show()
```

```python
marks=[45,55,60,55,60,66,78,66,98,88,66,98]
plt.hist(marks,bins=6,color="gold",edgecolor="red",label="marks")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Histogram")
plt.savefig("sample3.png")
plt.legend()
plt.show()
```
