```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[5,6,8,10,15]
x1=[2,4,6,8,10]
y1=[4,8,12,16,20]
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.xlim(1,10)
plt.ylim(5,15)
plt.title("Line Plot Example")
plt.plot(x,y,label="line-1",color='green',linestyle='dotted',linewidth=4,marker='D',markerfacecolor='purple',markersize=12)
plt.plot(x1,y1,label="line-2",color='brown',linestyle='dashed',linewidth=4,marker='*',markerfacecolor='orange',markersize=12,)
plt.legend()
plt.savefig("sample.png")
plt.show()
```
